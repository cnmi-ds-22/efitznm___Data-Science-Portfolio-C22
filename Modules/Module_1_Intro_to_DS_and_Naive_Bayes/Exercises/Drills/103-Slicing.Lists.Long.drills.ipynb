# List for Slicing Generator

The `gen_my_list()` function generates a list of random numbers that includes the words "dog", "cat", and "bat". These words appear in either forward or reverse order, with the same number of numbers separating each word.

Your task is four-fold:

1. pick a number for the seed between 0 and 49, inclusive
1. run the `gen_my_list()` function using the seed
1. display the following:
    - all or part of the list
    - the length of the list
    - the slice object or slice values ( i.e. start, stop, skip values ) for slicing out the words "dog", "cat", and "bat", in that order.
    - the slice of the list, which should output `['dog', 'cat', 'bat']`
1. repeat with a new seed, doing as many as you can in 5 minutes


```python
# @title Generator function

import random
def gen_my_list(seed: int) -> list:
    '''Provide an integer seed to randomly create a list of numbers'''
    random.seed(seed)
    prefix: list = list(range(4, random.randint(100,2000) ))
    random.shuffle(prefix)
    suffix: list = list(range(4, random.randint(100,2000) ))
    random.shuffle(suffix)
    tens: list = [10] * ( random.randint(100,500) * 2 + 1)
    tens[0::len(tens)//2] = ["dog","cat","bat"]
    tens = tens[::(random.randint(0,1)*2-1)]
    return prefix + tens + suffix
```

## Example with seed 0


```python
# generate the list
seed = 0
my_list = gen_my_list(seed)
```

```python
# Show the number of elements in the list
len(my_list)
```

```python
# View the list
", ".join([ str(x) for x in my_list ])
```

```python
# Slice the list for the target strings
my_list[1825:2286:230]
```
