## Processing



```python
# Split data frame into target and features
target = titanic_df['Survived']
features = titanic_df[['Age']] #[[because dataframe is two dimentional]] .shape gives you how many dimensions
```

```python
# Verify features
features
```

```python
# Verify target
target
```

### Single Cross Validation ( CV )

Run a single Cross Validation

#### Train Test Split

```python
# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.25)
```

```python
# Verify shape of each data set from Train Test Split
for i in [X_train, X_test, y_train, y_test]:
  display(i.shape)
```

#### Model


```python
# Create an empty GNB model
modelGNB = GaussianNB()
modelGNB
```

#### Fit



```python
# Train ( fit ) the model
modelGNB.fit((X_train), y_train)
```

#### Prediction


```python
# Make predicitons using the model
results = modelGNB.predict(X_test)
results
```

#### Performance


We'll meassure accuracy, which is the proportions of the number correctly predicted, which is mathematically the same as one minus the proportion incorrectly predicted.

```python
# Calculate the accuracy
acc = metrics.accuracy_score(y_test, results)
acc
```

```python
# Display a confusion matrix
matrix = metrics.confusion_matrix(y_test, results)

sns.heatmap(matrix.T, #T is shorthand for 'transpose'
  square=True,
  annot=True,
  fmt='d',
  cbar=False,
)

plt.xlabel('Actual')
plt.ylabel('Predicted')
```
